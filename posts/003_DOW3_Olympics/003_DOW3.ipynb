Since women's hockey officially became an Olympic sport in 1998, either USA or Canada has won each winter olympics cycle. 

### Olympic Women's Hockey Gold Medal Winners
- **2022** (Beijing): Canada 
- **2018** (Pyeongchang): USA 
- **2014** (Sochi): Canada 
- **2010** (Vancouver): Canada
- **2006** (Torino): Canada 
- **2002** (Salt Lake City): Canada 
- **1998** (Nagano): USA

**This made wonder what makes these two teams so dominant in ice hockey? And are there traits that can make up an ideal hockey player?**

```python
import pandas as pd
```

```python
medals_df = pd.read_csv('data/winter_olympics_medals.csv')
athletes_df = pd.read_csv('data/bios.csv')
```

```python
canada_gold = medals_df.query("event == 'Ice Hockey, Women (Olympic)' and team == 'Canada' and medal == 'Gold'")
```

```python
Canada_players = canada_gold["as"].drop_duplicates()
```

# The data does not point to a definitive answer of an ideal hockey player. However, it does reveal a consistent pattern: a significant number of Olympic athletes were born in hockey-heavy regions. For Canada, it was Ontario and Quebec. While in the United States, Massachusetts and Minnesota. And for Russia, Moskva. This might suggest that although a competitive athlete has to fit a certain physical frame, winning Gold may be less dependent on individual physical  variations and more about having accessibility to strong developmental programs, competitive youth leagues, and just the culture in general growing up.
