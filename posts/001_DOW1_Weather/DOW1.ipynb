## A) What was the average fahrenheit for the day? 

```python
birthday = weather_df.query('day == "January 30"') 
temp_avg = (birthday['high'] + birthday['low'])/2
weather_df = weather_df.assign(temp_avg=(weather_df['high']+weather_df['low'])/2)
```

```python
weather_df.query('day =="January 30"')
```

```python
birthday_mean_temp = temp_avg.mean()

print("Average birthday temperature:", round(birthday_mean_temp, 1), "degrees")
```

#### Since I was born in 2006, combining the weather data of Philadelphia for the past twenty years, the average fahrenheit on my birthday in Philadelphia is 32.3 degrees. 

Looking at the weather on my birthday from 2006 to 2025 showed that January 30 in Philadelphia is usually cold, but not always extremely cold. The average temperature across all years was about 32.3 degrees in Fahrenheit. While some birthdays were especially cold, others were surprisingly ok. Winter weather on my birthday was varied more than I expected.

## B) Did it rain? 

```python
birthday = weather_df.query('day == "January 30"') 
is_rain = birthday['rain']>0
```

```python
rain_counts = is_rain.value_counts()

rain_yes = rain_counts[True]
rain_no = rain_counts[False]
```

```python
print(f"Out of 21 birthdays analyzed, it rained on {rain_yes} birthdays and did not rain on {rain_no} birthdays.")
```

**Out of 21 birthdays, it rained on 8 birthdays and did not rain on 13 birthdays.**

Most of my birthdays in Philadelphia did not have rain. Around 62 percent out of all my birthdays recorded rainfall on January 30. Even through the winter, the data showed that rainy birthdays were actually less common for me.

## C) Did it snow? 

```python
birthday = weather_df.query('day == "January 30"') 
is_snow = birthday['snow']>0
```

```python
snow_counts = is_snow.value_counts()

snow_yes = snow_counts[True]
snow_no = snow_counts[False]
```

```python
print(f"Out of 21 birthdays analyzed, it snowed on {snow_yes} birthdays and did not snow on {snow_no} birthdays.")
```

**Out of 21 birthdays analyzed, it snowed on 7 birthdays and did not snow on 14 birthdays.**

Snowfall was also less common than I originally thought. Most of my birthdays did not have a lot of snowfall, even though January is in the winter. This was a suprising finding because January 30 is directly in the winter.
